### 🔧 Ejercicios (Funciones)
8. Escribir `es_palindromo(cadena)` que devuelva `True` si la cadena se lee igual al derecho y al revés (ignorar espacios y mayúsculas/minúsculas).
9. Escribir `contar_vocales(cadena)` que retorne un diccionario con la cuenta de cada vocal.
10. Escribir `normalizar_palabras(frase)` que retorne una **lista** de palabras sin signos de puntuación y en minúsculas.

```python
def es_palindromo(c: str) -> bool:
    """
    Determina si una cadena es un palíndromo.

    Parámetros:
        c (str): Cadena de texto ingresada por el usuario.

    Retorna:
        bool: True si la cadena se lee igual al derecho y al revés,
              False en caso contrario.

    Nota:
        - Se ignoran los espacios y las mayúsculas/minúsculas.
        - Ejemplo: "Anita lava la tina" → True
    """
    # Convertimos todo a minúsculas y eliminamos espacios
    a = c.lower().replace(" ", "")

    # Comparamos la cadena con su versión invertida
    return a == a[::-1]


# Programa principal
cadena = input("Ingrese una palabra o frase: ")
if es_palindromo(cadena):
    print("Es un palíndromo.")
else:
    print("No es un palíndromo.")
```

```python
def contar_vocales(c: str) -> dict:
    """
    Cuenta la cantidad de apariciones de cada vocal en una cadena.

    Parámetros:
        c (str): Cadena de texto ingresada por el usuario.

    Retorna:
        dict: Un diccionario con las vocales como claves y la cantidad de apariciones como valores.

    Ejemplo:
        contar_vocales("Hola") -> {'a': 1, 'e': 0, 'i': 0, 'o': 1, 'u': 0}
    """
    c = c.lower()  # Pasamos todo a minúsculas
    vocales = {"a": 0, "e": 0, "i": 0, "o": 0, "u": 0}  # Diccionario base

    # Recorremos la cadena letra por letra
    for car in c:
        if car in vocales:
            vocales[car] += 1

    return vocales


# Programa principal
palabra = input("Ingrese una palabra: ")
print(contar_vocales(palabra))
```

```python
import string

def normalizar_palabras(c: str) -> list:
    """
    Normaliza una frase eliminando signos de puntuación
    y convirtiendo todo el texto a minúsculas.

    Parámetros:
        c (str): Frase o texto completo a limpiar.

    Retorna:
        list: Lista de palabras limpias (sin puntuación ni mayúsculas).

    Ejemplo:
        normalizar_palabras("Hola, mundo!") -> ['hola', 'mundo']
    """
    c = c.lower()  # Convertimos a minúsculas

    # Recorremos todos los signos de puntuación estándar y los eliminamos
    for signo in string.punctuation:
        c = c.replace(signo, "")

    # Dividimos en palabras separadas por espacios
    return c.split()


# Programa principal
frase = input("Escriba una frase: ")
print(normalizar_palabras(frase))
```

### 🔧 Ejercicios (Listas)
11. Ingresar notas hasta `-1`, calcular el **promedio** y cuántos están **por debajo**.
12. Dada una lista de palabras, construir otra con las que tengan **más de 5** letras.

```python
def analizar_notas() -> None:
    """
    Lee notas ingresadas por el usuario hasta que se introduzca -1.
    Luego calcula el promedio general y cuántas notas están por debajo de ese promedio.

    No recibe parámetros.
    No retorna nada, solo imprime los resultados.

    Ejemplo de ejecución:
        Ingrese nota: 7
        Ingrese nota: 4
        Ingrese nota: 9
        Ingrese nota: -1
        El promedio es 6.67
        Hay 1 notas menores al promedio
    """
    print("Ingrese notas o -1 para terminar")

    # Inicialización de variables
    notas = int(input("Ingrese nota: "))
    lista_notas = []
    suma = 0
    conteo = 0

    # Bucle de carga de notas
    while notas != -1:
        lista_notas.append(notas)
        suma += notas
        conteo += 1
        notas = int(input("Ingrese nota: "))

    # Procesamiento y salida
    if conteo > 0:
        promedio = suma / conteo
        debajo = 0
        for n in lista_notas:
            if n < promedio:
                debajo += 1
        print(f"\nEl promedio es {promedio:.2f}")
        print(f"Hay {debajo} notas menores al promedio")
    else:
        print("No se ingresaron notas válidas.")


# Programa principal
analizar_notas()
```

```python
def filtrar_palabras_largas(lista: list) -> list:
    """
    Recibe una lista de palabras y devuelve otra lista
    con aquellas que tengan más de 5 letras.

    Parámetros:
        lista (list): Lista de palabras (strings).

    Retorna:
        list: Nueva lista con palabras de longitud mayor a 5.

    Ejemplo:
        filtrar_palabras_largas(["cosa", "solecitos", "sol"])
        -> ['solecitos']
    """
    lista2 = []
    for palabra in lista:
        if len(palabra) > 5:
            lista2.append(palabra)
    return lista2


# Programa principal
lista = ["cosa", "cosas", "sol", "solecitos"]
print(f"Palabras con más de 5 letras: {filtrar_palabras_largas(lista)}")
```

### 🔧 Ejercicios (Tuplas)
13. Guardar coordenadas `(x, y)` en una tupla e imprimir la distancia al origen.
14. Dada una lista de tuplas `(nombre, edad)`, imprimir solo los mayores de 18.

```python
import math

def distancia_al_origen(x: int, y: int) -> float:
    """
    Calcula la distancia de un punto (x, y) al origen (0, 0).

    Parámetros:
        x (int): Coordenada en el eje X.
        y (int): Coordenada en el eje Y.

    Retorna:
        float: Distancia al origen calculada con la fórmula √(x² + y²).

    Ejemplo:
        distancia_al_origen(3, 4) -> 5.0
    """
    return math.sqrt(x**2 + y**2)


# Programa principal
x = int(input("Ingrese una coordenada en x: "))
y = int(input("Ingrese una coordenada en y: "))

tupla = (x, y)  # Guardamos las coordenadas en una tupla
distancia = distancia_al_origen(*tupla)  # Desempaquetamos la tupla al llamar la función

print(f"La distancia al origen es {distancia:.2f}")
```

```python
def mostrar_mayores(lista: list) -> None:
    """
    Imprime los nombres de las personas mayores de 18 años
    a partir de una lista de tuplas (nombre, edad).

    Parámetros:
        lista (list): Lista de tuplas, donde cada tupla contiene (nombre, edad).

    Retorna:
        None. Solo imprime resultados.

    Ejemplo:
        mostrar_mayores([("Ana", 23), ("Luis", 16)]) -> Imprime "Ana"
    """
    for nombre, edad in lista:
        if edad > 18:
            print(nombre)


# Programa principal
personas = [("a", 23), ("b", 16), ("c", 20)]
print("Mayores de edad:")
mostrar_mayores(personas)
```

```python
def procesar_notas() -> None:
    """
    Lee nombres y notas desde teclado y los guarda en un diccionario.
    Luego calcula y muestra:
        - El promedio general.
        - El mejor estudiante.
        - Los que tienen nota por debajo del promedio.

    Retorna:
        None (solo imprime los resultados).

    Ejemplo:
        Entrada:
            Ana 8
            Luis 5
            Carla 9
            FIN
        Salida:
            Promedio general: 7.33
            Carla tiene la mejor nota con 9
            Luis tiene una nota menor al promedio
    """

    pares = {}
    nombres = input("Teclee nombre o FIN para finalizar: ")

    # Carga de datos en el diccionario
    while nombres != "FIN":
        notas = int(input(f"Nota de {nombres}: "))
        pares[nombres] = notas
        nombres = input("Teclee nombre o FIN para finalizar: ")

    if not pares:
        print("No se ingresaron datos.")
        return

    # Promedio general
    promedio = sum(pares.values()) / len(pares)
    print(f"Promedio general: {promedio:.2f}")

    # Mejor estudiante
    mejor_estudiante = max(pares, key=pares.get)
    print(f"{mejor_estudiante} tiene la mejor nota con {pares[mejor_estudiante]}")

    # Estudiantes debajo del promedio
    for nombre, nota in pares.items():
        if nota < promedio:
            print(f"{nombre} tiene una nota menor al promedio")


# Programa principal
procesar_notas()
```

```python
def palabras_longitud(frase: str) -> dict:
    """
    Construye un diccionario donde cada palabra es una clave
    y su valor asociado es la cantidad de letras que tiene.

    Parámetros:
        frase (str): Texto ingresado por el usuario.

    Retorna:
        dict: Diccionario con la forma {palabra: longitud}.

    Ejemplo:
        Entrada: "hola mundo"
        Salida: {"hola": 4, "mundo": 5}
    """
    palabras = frase.split()
    diccionario = {}

    for palabra in palabras:
        diccionario[palabra] = len(palabra)

    return diccionario


# Programa principal
frase = input("Escriba una frase: ")
print(palabras_longitud(frase))
```

### 🔧 Ejercicios (Sets)
17. Pedir dos listas de números (separados por espacios) y mostrar su **intersección** usando `set`.
18. Eliminar duplicados de una lista manteniendo el **orden** (pista: usar un `set` auxiliar para vistos).

```python
def interseccion_listas(lista1: list, lista2: list) -> set:
    """
    Calcula la intersección entre dos listas de números usando conjuntos (set).
    La intersección contiene solo los elementos que aparecen en ambas listas.

    Parámetros:
        lista1 (list): Primera lista de números.
        lista2 (list): Segunda lista de números.

    Retorna:
        set: Conjunto con los elementos comunes a ambas listas.

    Ejemplo:
        interseccion_listas([1, 2, 3], [2, 3, 4]) -> {2, 3}
    """
    return set(lista1) & set(lista2)


# Programa principal
lista1 = []
for i in range(3):
    num = int(input("Ingrese un número para la lista 1: "))
    lista1.append(num)

lista2 = []
for i in range(3):
    num = int(input("Ingrese un número para la lista 2: "))
    lista2.append(num)

# Mostrar la intersección
print("Intersección:", interseccion_listas(lista1, lista2))
```

```python
def eliminar_duplicados(lista: list) -> list:
    """
    Elimina los elementos duplicados de una lista manteniendo su orden original.

    Parámetros:
        lista (list): Lista que puede contener elementos repetidos.

    Retorna:
        list: Nueva lista sin duplicados, en el mismo orden en que aparecieron.

    Ejemplo:
        eliminar_duplicados([1, 2, 2, 3, 3, 4]) -> [1, 2, 3, 4]
    """
    vistos = set()          # conjunto para registrar los elementos ya encontrados
    sin_duplicados = []     # lista final sin repetidos

    for num in lista:
        if num not in vistos:
            sin_duplicados.append(num)
            vistos.add(num)

    return sin_duplicados


# Programa principal
lista = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]
print("Lista original:", lista)
print("Sin duplicados:", eliminar_duplicados(lista))
```

```python
def frecuencias_letras(frase: str) -> dict:
    """
    Calcula la frecuencia de aparición de cada letra en una frase.
    Ignora los espacios y no distingue entre mayúsculas y minúsculas.

    Parámetros:
        frase (str): Cadena de texto a analizar.

    Retorna:
        dict: Diccionario {letra: cantidad_de_apariciones}.

    Ejemplo:
        frecuencias_letras("Hola Mundo") -> {'h':1, 'o':2, 'l':1, 'a':1, 'm':1, 'u':1, 'n':1, 'd':1}
    """
    frecuencias = {}
    for letra in frase.lower():
        if letra != " ":
            frecuencias[letra] = frecuencias.get(letra, 0) + 1
    return frecuencias


# Programa principal
frase = input("Ingrese una frase: ")
print(frecuencias_letras(frase))
```

```python
def analizar_notas() -> None:
    """
    Lee pares (nombre, nota) hasta que se ingresa 'FIN' y luego muestra:
    - Promedio general
    - Mediana
    - Mejor y peor nota
    - Informe ordenado alfabéticamente por nombre
    """
    notas = []

    nombre = input("Ingrese nombre o FIN para terminar: ")
    while nombre != "FIN":
        nota = int(input(f"Ingrese nota de {nombre}: "))
        notas.append((nombre, nota))
        nombre = input("Ingrese nombre o FIN para terminar: ")

    if not notas:
        print("No se ingresaron notas.")
        return

    # Promedio
    promedio = sum(n[1] for n in notas) / len(notas)
    print(f"\nPromedio general: {promedio:.2f}")

    # Mediana
    notas_ordenadas = sorted(notas, key=lambda x: x[1])
    n = len(notas_ordenadas)
    if n % 2 == 1:
        mediana = notas_ordenadas[n // 2][1]
    else:
        mediana = (notas_ordenadas[n // 2 - 1][1] + notas_ordenadas[n // 2][1]) / 2
    print(f"Mediana: {mediana:.2f}")

    # Mejor y peor
    mejor = max(notas, key=lambda x: x[1])
    peor = min(notas, key=lambda x: x[1])
    print(f"Mejor nota: {mejor[0]} ({mejor[1]})")
    print(f"Peor nota: {peor[0]} ({peor[1]})")

    # Informe ordenado por nombre
    print("\nInforme ordenado por nombre:")
    for nombre, nota in sorted(notas, key=lambda x: x[0].lower()):
        print(f"{nombre}: {nota}")


# Programa principal
analizar_notas()
```

```python
def letras_unicas(frase: str) -> set:
    """
    Devuelve el conjunto de letras únicas (sin espacios, minúsculas).

    Parámetros:
        frase (str): Texto a analizar.

    Retorna:
        set: Conjunto con las letras únicas.

    Ejemplo:
        letras_unicas("Hola mundo") -> {'h', 'o', 'l', 'a', 'm', 'u', 'n', 'd'}
    """
    return set(frase.replace(" ", "").lower())


# Programa principal
frase = input("Ingrese una frase: ")
letras = letras_unicas(frase)

print(f"Letras únicas ({len(letras)}): {letras}")
print("Lista ordenada:", sorted(list(letras)))
```
